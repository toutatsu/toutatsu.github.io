```python
import os
import csv
```

```python
for dirpath, dirnames, filenames in os.walk('.'):
    if "./.git" in dirpath:continue
    dirnames=[dir for dir in dirnames if dir[0]!='.']
    filenames=[file for file in filenames if file[0]!='.' and file[-5:]=='.html']
    print(f"{dirpath=}")
    print(f"{dirnames=}")
    print(f"{filenames=}\n\n")

    with open(dirpath+'/contents.csv', 'w', newline='') as contents_file:
        writer = csv.writer(contents_file)
        writer.writerow([dirpath])
        writer.writerow(dirnames)
        writer.writerow(filenames)

    # os.remove(dirpath+'/contents.csv')
```
